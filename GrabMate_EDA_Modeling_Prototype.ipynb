```python
# ============================================================
# CELL 1 – TỔNG HỢP EDA
# 1A. EDA traffic  – grab_traffic_segments
# 1B. EDA multimodal trips (V17 mục 1→4B)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cố gắng import folium (vẽ map); nếu không có thì bỏ qua phần map
try:
    import folium
    from IPython.display import display
except ImportError:
    folium = None
    def display(x): print(x)

# ------------------------------------------------------------
# CONFIG CHUNG
# ------------------------------------------------------------
FILE_PATH = "data/hcm_multimodal_dataset_v3.xlsx"
TRAFFIC_SHEET = "grab_traffic_segments"
TRIP_SHEET    = "multimodal_trip_samples"
TARGET_COL    = "recommended_total_cost_vnd"
RANDOM_STATE  = 42

# ============================================================
# 1A. EDA – GRAB TRAFFIC SEGMENTS (Tâm)
# ============================================================

grab_seg = pd.read_excel(FILE_PATH, sheet_name=TRAFFIC_SHEET).copy()
print("grab_seg shape:", grab_seg.shape)
display(grab_seg.head())

# --- 1A.1. Biểu đồ theo giờ & traffic_level ---
grab_seg['timestamp'] = pd.to_datetime(grab_seg['timestamp'], format='%H:%M')
# giả lập ngày cụ thể
grab_seg['datetime'] = pd.to_datetime('2025-11-16') + pd.to_timedelta(
    grab_seg['timestamp'].dt.strftime('%H:%M:%S')
)
grab_seg['hour'] = grab_seg['datetime'].dt.hour

plt.figure(figsize=(16, 5))

# Countplot: số đoạn đường theo traffic_level & giờ
plt.subplot(1, 3, 1)
sns.countplot(
    data=grab_seg, x='hour', hue='traffic_level',
    hue_order=['low', 'medium', 'high'], palette='Reds_r'
)
plt.title('Số đoạn đường theo traffic_level & giờ')
plt.xlabel('Giờ trong ngày')
plt.legend(title='Mức kẹt')

# % đoạn high traffic theo giờ
plt.subplot(1, 3, 2)
high_pct = grab_seg[grab_seg['traffic_level'] == 'high'].groupby('hour').size() / grab_seg.groupby('hour').size()
high_pct = high_pct.fillna(0)
high_pct.plot(kind='bar', color='darkred', alpha=0.8)
plt.title('% đoạn đường HIGH TRAFFIC')
plt.ylabel('Tỷ lệ')
plt.xlabel('Giờ')

# Boxplot: active_vehicles theo giờ
plt.subplot(1, 3, 3)
sns.boxplot(data=grab_seg, x='hour', y='active_vehicles', palette='Oranges')
plt.title('Số xe hoạt động theo giờ')
plt.xlabel('Giờ')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

print("INSIGHT (traffic theo giờ): Kẹt xe & số xe hoạt động tăng mạnh vào các khung 7h, 12h, 17h, 21h (tùy dữ liệu).")

# --- 1A.2. Bản đồ hotspot kẹt xe ---
if folium is not None:
    m = folium.Map(location=[10.7769, 106.7009], zoom_start=12, tiles='CartoDB positron')

    color_map = {'low': 'green', 'medium': 'orange', 'high': 'red'}

    for _, row in grab_seg.iterrows():
        folium.CircleMarker(
            location=[row['latitude'], row['longitude']],
            radius=row['active_vehicles'] / 3,
            color=color_map[row['traffic_level']],
            fill=True,
            fill_color=color_map[row['traffic_level']],
            fill_opacity=0.7,
            popup=folium.Popup(
                f"<b>{row['road_name']}</b><br>"
                f"Thời gian: {row['timestamp'].strftime('%H:%M')}<br>"
                f"Xe: {row['active_vehicles']}<br>"
                f"Mức kẹt: {row['traffic_level'].upper()}",
                max_width=300
            )
        ).add_to(m)

    legend_html = '''
    <div style="position: fixed; bottom: 50px; left: 50px; z-index: 1000;
                background-color: white; padding: 10px; border-radius: 5px;
                border: 1px solid gray; font-size: 12px;">
        <p><b>Mức kẹt xe</b></p>
        <p><span style="color:green;">●</span> Low</p>
        <p><span style="color:orange;">●</span> Medium</p>
        <p><span style="color:red;">●</span> High</p>
    </div>
    '''
    m.get_root().html.add_child(folium.Element(legend_html))

    display(m)
    m.save("traffic_hotspots_folium.html")
    print("ĐÃ LƯU: traffic_hotspots_folium.html")
else:
    print("Không có folium – bỏ qua map traffic.")

# --- 1A.3. Validation: active_vehicles vs traffic_level ---
print(f"\nĐang validate trên {len(grab_seg)} đoạn đường...")

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
order = ['low', 'medium', 'high']
sns.boxplot(
    data=grab_seg, x='traffic_level', y='active_vehicles',
    order=order, palette='Set1'
)
plt.title('Phân bố active_vehicles theo traffic_level')
plt.ylabel('Số xe hoạt động')
plt.xlabel('Mức kẹt xe')

plt.subplot(1, 2, 2)
level_map = {'low': 1, 'medium': 2, 'high': 3}
grab_seg['level_num'] = grab_seg['traffic_level'].map(level_map)

sns.regplot(
    data=grab_seg,
    x='active_vehicles', y='level_num',
    scatter_kws={'alpha': 0.7, 's': 60},
    line_kws={'color': 'red', 'linewidth': 2}
)
plt.yticks([1, 2, 3], ['low', 'medium', 'high'])
plt.title('Mối quan hệ: active_vehicles ↑ → traffic_level ↑')
plt.xlabel('Số xe hoạt động')
plt.ylabel('Mức kẹt xe')

plt.tight_layout()
plt.show()

corr = grab_seg['active_vehicles'].corr(grab_seg['level_num'])
print(f"\nCorrelation (Pearson): {corr:.3f}")
if corr > 0.7:
    print("→ MẠNH: Dữ liệu traffic validate rất tốt.")
elif corr > 0.4:
    print("→ TRUNG BÌNH: Có mối quan hệ.")
else:
    print("→ YẾU: Cần kiểm tra lại logic data traffic.")

print("\nINSIGHT VALIDATION: active_vehicles càng cao thì traffic_level có xu hướng cao hơn → dữ liệu hợp lý để dùng làm feature.")

# ============================================================
# 1B. EDA – MULTIMODAL TRIP SAMPLES (Nam – V17 mục 1→4B)
# ============================================================

df = pd.read_excel(FILE_PATH, sheet_name=TRIP_SHEET)
df_raw = df.copy()
print("\n===== MULTIMODAL TRIP SAMPLES =====")
print("Kích thước dữ liệu:", df.shape)
display(df.head())
print("\nThông tin dữ liệu:")
df.info()

# --- 2. Missing values ---
print("\n===== Missing Values (SL) =====")
print(df.isna().sum().sort_values(ascending=False))

print("\n===== Missing Values (%) =====")
print((df.isna().mean() * 100).sort_values(ascending=False))

# --- 3. Histograms ---
candidate_cols = [
    "distance_km",
    TARGET_COL,
    "total_time_min",
    "grab_estimated_cost_vnd",
    "recommended_total_time_min"
]
print("\n===== Vẽ Histograms =====")
for col in candidate_cols:
    if col in df.columns:
        plt.figure()
        df[col].hist(bins=50)
        plt.title(f"Histogram of {col}")
        plt.xlabel(col)
        plt.ylabel("Count")
        plt.tight_layout()
        plt.show()

# --- 3A. Correlation heatmap ---
eda_corr_cols = [
    TARGET_COL,
    "distance_km",
    "recommended_total_time_min",
    "walking_distance_to_transit_m",
    "duration_min_grab_only",
    "total_cost_grab_only_vnd",
    "duration_min_grab_bus",
    "total_cost_grab_bus_vnd",
    "duration_min_grab_metro",
    "total_cost_grab_metro_vnd",
    "passenger_count",
    "bus_num_stops",
    "metro_num_stations",
    "grab_eta_to_user_min",
    "nearest_driver_distance_km"
]
existing_eda_cols = [c for c in eda_corr_cols if c in df.columns]

if len(existing_eda_cols) > 1:
    plt.figure(figsize=(14, 10))
    corr_matrix = df[existing_eda_cols].corr()
    sns.heatmap(
        corr_matrix,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0,
        annot_kws={"size": 8}
    )
    plt.title("Ma trận tương quan giữa các biến (EDA – multimodal)")
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
else:
    print("Không đủ biến numeric để vẽ heatmap tương quan.")

# --- 3B. Map điểm bắt đầu ---
if folium is not None:
    print("\n===== EDA MAP PLOT – Điểm xuất phát =====")
    lat_col = None
    lon_col = None
    for c in ["origin_lat", "start_lat", "pickup_latitude"]:
        if c in df_raw.columns:
            lat_col = c
            break
    for c in ["origin_lon", "start_lon", "pickup_longitude"]:
        if c in df_raw.columns:
            lon_col = c
            break

    if lat_col and lon_col:
        sample_map = df_raw.sample(min(500, df_raw.shape[0]), random_state=RANDOM_STATE)
        center_lat = sample_map[lat_col].mean()
        center_lon = sample_map[lon_col].mean()
        m_eda = folium.Map(location=[center_lat, center_lon], zoom_start=11)
        for _, row in sample_map.iterrows():
            folium.CircleMarker(
                location=[row[lat_col], row[lon_col]],
                radius=2, color="blue", fill=True, fill_opacity=0.7
            ).add_to(m_eda)
        display(m_eda)
        m_eda.save("eda_pickup_points_map.html")
        print("Đã lưu map EDA tại: eda_pickup_points_map.html")
    else:
        print("Không có lat/lon để vẽ map EDA.")
else:
    print("Không vẽ được map EDA vì thiếu folium.")

# --- 4. Outliers + lọc dữ liệu vô lý ---
def detect_iqr(series, k=1.5):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    return Q1 - k * IQR, Q3 + k * IQR

print("\n===== Phân tích Outliers (chỉ kiểm tra, không xoá IQR) =====")
check_cols = ["distance_km", TARGET_COL]
for col in check_cols:
    if col in df.columns:
        lo, up = detect_iqr(df[col])
        rate = ((df[col] < lo) | (df[col] > up)).mean() * 100
        print(f"{col}: lower={lo:.3f}, upper={up:.3f}, outlier rate={rate:.2f}%")

print("\nLọc dữ liệu vô lý (distance_km, cost, total_time_min > 0)")
before_count = len(df)
if "distance_km" in df.columns:
    df = df[df["distance_km"] > 0]
if TARGET_COL in df.columns:
    df = df[df[TARGET_COL] > 0]
if "total_time_min" in df.columns:
    df = df[df["total_time_min"] > 0]
after_count = len(df)
print(f"Kích thước dữ liệu sau khi lọc: {before_count} -> {after_count}")

# --- 4B. Map vẽ đường đi (routes) ---
if folium is not None:
    print("\n===== Vẽ Map tuyến đường (routes) =====")
    lat_start_col = lon_start_col = lat_end_col = lon_end_col = None

    if "start_lat" in df.columns and "start_lon" in df.columns:
        lat_start_col, lon_start_col = "start_lat", "start_lon"
    elif "origin_lat" in df.columns and "origin_lon" in df.columns:
        lat_start_col, lon_start_col = "origin_lat", "origin_lon"

    if "end_lat" in df.columns and "end_lon" in df.columns:
        lat_end_col, lon_end_col = "end_lat", "end_lon"
    elif "destination_lat" in df.columns and "destination_lon" in df.columns:
        lat_end_col, lon_end_col = "destination_lat", "destination_lon"

    if lat_start_col and lon_start_col and lat_end_col and lon_end_col and 'recommended_route_type' in df.columns:
        sample_routes = df.sample(min(150, df.shape[0]), random_state=RANDOM_STATE)
        print(f"Đang vẽ {len(sample_routes)} hành trình mẫu...")

        m_routes = folium.Map(
            location=[sample_routes[lat_start_col].mean(), sample_routes[lon_start_col].mean()],
            zoom_start=11
        )

        color_map_routes = {
            'grab_only': '#E74C3C',
            'grab_bus_grab': '#3498DB',
            'grab_metro_grab': '#9B59B6',
            'walk_bus_walk': '#2ECC71',
            'walk_metro_walk': '#F1C40F'
        }

        legend_html = """
         <div style="position: fixed; 
         bottom: 50px; left: 50px; width: 170px; height: 130px; 
         background-color: rgba(255, 255, 255, 0.8);
         border:2px solid grey; z-index:9999; font-size:12px;
         border-radius: 10px; padding: 5px;">
         <b>Chú thích</b><br>
         <span style="color:#E74C3C;">▬</span> Grab Only<br>
         <span style="color:#3498DB;">▬</span> Grab + Bus<br>
         <span style="color:#9B59B6;">▬</span> Grab + Metro<br>
         <span style="color:#2ECC71;">▬</span> Walk + Bus<br>
         <span style="color:#F1C40F;">▬</span> Walk + Metro<br>
         </div>
        """
        m_routes.get_root().html.add_child(folium.Element(legend_html))

        for _, row in sample_routes.iterrows():
            start_coords = [row[lat_start_col], row[lon_start_col]]
            end_coords = [row[lat_end_col], row[lon_end_col]]
            route_type = row.get('recommended_route_type', 'grab_only')
            color = color_map_routes.get(route_type, '#95A5A6')

            popup_text = (
                f"<b>Loại:</b> {route_type}<br>"
                f"<b>Chi phí (Target):</b> {row[TARGET_COL]:,.0f} VND<br>"
                f"<b>Thời gian:</b> {row['recommended_total_time_min']:.1f} phút"
            )

            folium.PolyLine(
                [start_coords, end_coords],
                color=color,
                weight=1.5,
                opacity=0.7,
                popup=folium.Popup(popup_text, max_width=200)
            ).add_to(m_routes)

            folium.CircleMarker(
                start_coords,
                radius=1,
                color=color,
                fill=True,
                fill_opacity=1
            ).add_to(m_routes)

        display(m_routes)
        m_routes.save("trip_routes_map_colored.html")
        print("Đã lưu bản đồ đường đi tại: trip_routes_map_colored.html")
    else:
        print("Không đủ cột lat/lon hoặc recommended_route_type để vẽ map routes.")
else:
    print("Không vẽ được map routes vì thiếu folium.")

print("\n=== CELL 1 (EDA) HOÀN TẤT ===")
```

```python
# ============================================================
# HCM MULTIMODAL FARE – FULL PIPELINE (PHIÊN BẢN 17)
#
# PHIÊN BẢN 17 (CHUẨN - SỬA LEAKAGE TRIỆT ĐỂ):
# 1. Mục 7: Xóa TẤT CẢ features gây leakage (time, duration, stops...)
# 2. Mục 5 & 7: Chuyển Imputer vào BÊN TRONG Pipeline
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from math import radians, sin, cos, asin, sqrt

# Cần thư viện này để hiển thị map và bảng kết quả đẹp
try:
    import folium
    from IPython.display import display, HTML
except ImportError:
    print("Cảnh báo: Cần cài đặt 'folium' và 'ipython' để vẽ map và hiển thị bảng kết quả đẹp.")
    # Định nghĩa hàm display giả để code không lỗi nếu chạy ở môi trường không có IPython
    def display(x):
        print(x)

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer # Sẽ dùng trong Mục 7

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 0. CONFIG
# ------------------------------------------------------------

print("="*50)
print("BẠN ĐANG CHẠY FILE V17 (PHIÊN BẢN CHUẨN - ĐÃ SỬA LEAKAGE TRIỆT ĐỂ)")
print("Nếu thấy dòng này, kết quả R2 thật sẽ khoảng ~0.62")
print("="*50)


FILE_PATH = "data/hcm_multimodal_dataset_v3.xlsx"
MAIN_SHEET = "multimodal_trip_samples"

TARGET_COL = "recommended_total_cost_vnd"
RANDOM_STATE = 42

def compute_rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------
df = pd.DataFrame() # Khởi tạo df rỗng
try:
    df = pd.read_excel(FILE_PATH, sheet_name=MAIN_SHEET)
except FileNotFoundError:
    print(f"LỖI: Không tìm thấy file tại '{FILE_PATH}'.")
    print("Vui lòng cập nhật biến FILE_PATH ở mục 0. CONFIG")
except Exception as e:
    print(f"Lỗi khi đọc file: {e}")

if df.empty:
    print("Không thể tải dữ liệu. Dừng thực thi.")
    exit()
else:
    df_raw = df.copy()  # để vẽ map EDA
    print("Kích thước dữ liệu:", df.shape)
    print("\n5 dòng đầu:")
    display(df.head())
    print("\nThông tin dữ liệu:")
    df.info()


# ------------------------------------------------------------
# 2. IMPUTATION 
# ------------------------------------------------------------

print("\n===== Xử lý Missing Values =====")
print("Lưu ý: Imputation (Median/Mode) sẽ được thực hiện BÊN TRONG Pipeline (Mục 7)")
print("để tránh rò rỉ dữ liệu từ tập Test. Mục 5 này sẽ được bỏ qua.")



# ------------------------------------------------------------
# 3. FEATURE ENGINEERING 
# ------------------------------------------------------------
print("\n===== Feature Engineering =====")

def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    dlat = radians(lat2 - lat1); dlon = radians(lon2 - lon1)
    lat1 = radians(lat1); lat2 = radians(lat2)
    a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    c = 2 * asin(sqrt(a))
    return R * c

if "distance_km" not in df.columns:
    if all(c in df.columns for c in ["origin_lat", "origin_lon", "destination_lat", "destination_lon"]):
        df["distance_km"] = df.apply(
            lambda r: haversine(
                r["origin_lat"], r["origin_lon"],
                r["destination_lat"], r["destination_lon"]
            ), axis=1)
        print("Đã tính thêm distance_km bằng Haversine.")
    else:
        print("Không đủ cột để tính distance_km; bỏ qua bước này.")

# Dùng 'time_of_day_slot' có sẵn, không cần tạo 'hour'
print("Sử dụng 'time_of_day_slot' có sẵn làm feature thời gian.")


# ------------------------------------------------------------
# 4. MODEL DATA PREP 
# ------------------------------------------------------------

# === 1. ĐỊNH NGHĨA FEATURES (ĐÃ SỬA LEAKAGE) ===
potential_features = [
    # === Features "THÔ" (An toàn, KHÔNG leakage) ===
    # Đây là những thông tin bạn biết TRƯỚC KHI tìm đường
    
    "distance_km",          # Khoảng cách chim bay
    "traffic_level",        # Tình trạng giao thông
    "bus_available",        # Khu vực này có bus không
    "metro_available",      # Khu vực này có metro không
    "passenger_count",      # Số hành khách
    "time_of_day_slot",     # Khung giờ
    
    # (Các feature "xung quanh", tạm chấp nhận)
    "walking_distance_to_transit_m", 
    "grab_eta_to_user_min",          
    "nearest_driver_distance_km"     
    
    
]

# Chỉ lấy các cột thực sự tồn tại trong DataFrame
feature_cols = [col for col in potential_features if col in df.columns]

# *** KIỂM TRA LEAKAGE CUỐI CÙNG ***
if TARGET_COL in feature_cols:
    print(f"LỖI NGHIÊM TRỌNG: TARGET '{TARGET_COL}' đang nằm trong feature_cols!")
    print("Xóa target ra khỏi features...")
    feature_cols.remove(TARGET_COL)

print(f"\n===== Chuẩn bị dữ liệu Model (với {len(feature_cols)} features) =====")
print("Các features được sử dụng (ĐÃ SỬA LEAKAGE TRIỆT ĐỂ):", feature_cols)

X = df[feature_cols]
y = df[TARGET_COL]

num_features = [c for c in feature_cols if pd.api.types.is_numeric_dtype(X[c])]
cat_features = [c for c in feature_cols if not pd.api.types.is_numeric_dtype(X[c])]

print("\nNumeric features:", num_features)
print("Categorical features:", cat_features) 

# === 2. TẠO PIPELINE TIỀN XỬ LÝ (ĐÃ BAO GỒM IMPUTER) ===
# (Đây là best practice, thay thế cho Mục 5)

numeric_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')), # Thêm Imputer
    ('scaler', StandardScaler())                   # Scaler
])

categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')), # Thêm Imputer
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False)) # OneHot
])

# 3. Kết hợp chúng trong ColumnTransformer
preprocess = ColumnTransformer([
    ("num", numeric_pipeline, num_features),     # Dùng pipeline con
    ("cat", categorical_pipeline, cat_features)  # Dùng pipeline con
])

# === 4. CHIA TẬP TRAIN/TEST ===
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

print("\nTrain/Test:", X_train.shape, X_test.shape)

# ------------------------------------------------------------
# 5. MODELING - (KHÔNG THAY ĐỔI)
# ------------------------------------------------------------

models = {
    "Dummy": DummyRegressor(strategy="mean"),
    "LinearRegression": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "DecisionTree_Simple": DecisionTreeRegressor(
        max_depth=6, min_samples_leaf=50, random_state=RANDOM_STATE
    ),
    "RandomForest_Deep": RandomForestRegressor(
        n_estimators=100,
        max_depth=None,
        min_samples_leaf=2,
        n_jobs=-1,
        random_state=RANDOM_STATE
    )
}

results = []
y_pred_all = {}

for name, model in models.items():
    print(f"\n===== TRAINING {name} =====")
    # LƯU Ý: Pipeline giờ đã bao gồm Imputer, Scaler, OneHot và Model
    pipe = Pipeline([("prep", preprocess), ("model", model)])
    
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    y_pred_all[name] = pred
    
    results.append({
        "model": name,
        "MAE": mean_absolute_error(y_test, pred),
        "MAPE(%)": np.mean(np.abs((y_test - pred) / np.maximum(y_test, 1))) * 100,
        "RMSE": compute_rmse(y_test, pred),
        "R2": r2_score(y_test, pred)
    })

results_df = pd.DataFrame(results).sort_values("MAE")

print("\n===== KẾT QUẢ MÔ HÌNH (ĐÃ SỬA LEAKAGE TRIỆT ĐỂ) =====")
try:
    results_df_styled = results_df.style.format({
        "MAE": "{:,.2f}", "MAPE(%)": "{:,.2f}%", "RMSE": "{:,.2f}", "R2": "{:,.3f}"
    }).set_caption("Bảng so sánh kết quả các mô hình (Đã sửa Leakage Triệt Để V17)") \
      .set_properties(**{'text-align': 'center', 'width': '120px'}) \
      .background_gradient(cmap='viridis_r', subset=['MAE', 'MAPE(%)', 'RMSE']) \
      .highlight_max(subset=["R2"], color="#90EE90")
    display(results_df_styled)
except Exception as e:
    print(f"Lỗi styling DataFrame (bỏ qua, in bình thường): {e}")
    print(results_df)

# ============================================================
# 6. MODEL VISUAL EVALUATION
# ============================================================

print("\n===== Đánh giá mô hình trực quan (Đã sửa Leakage Triệt Để) =====")

# 7.1. Heatmap so sánh mô hình
plt.figure(figsize=(10, 6))
sns.heatmap(
    results_df.set_index("model")[["MAE", "MAPE(%)", "RMSE", "R2"]],
    annot=True, fmt=".3f", cmap="viridis_r"
)
plt.title("So sánh hiệu suất các mô hình (Đã sửa Leakage Triệt Để)")
plt.tight_layout()
plt.show()

# 7.2. ACTUAL vs PREDICTED – Top 3 mô hình theo RMSE
top3 = results_df[results_df['model'] != 'Dummy'].nsmallest(3, "RMSE").copy()
if not top3.empty:
    plt.figure(figsize=(18, 5))
    for i, row in enumerate(top3.itertuples(), start=1):
        name = row.model; rmse = row.RMSE; y_pred = y_pred_all[name]
        indices = np.random.choice(len(y_test), min(500, len(y_test)), replace=False)
        plt.subplot(1, 3, i)
        plt.scatter(y_test.iloc[indices], y_pred[indices], alpha=0.6)
        min_val = min(y_test.min(), y_pred.min()); max_val = max(y_test.max(), y_pred.max())
        plt.plot([min_val, max_val], [min_val, max_val], "k--", label="Ideal")
        plt.xlabel("Actual Price");
        if i == 1: plt.ylabel("Predicted Price")
        plt.title(f"{name} | RMSE={rmse:.0f}")
        plt.grid(True); plt.legend()
    plt.suptitle("Actual vs Predicted - Top 3 Models (theo RMSE - Đã sửa Leakage Triệt Để)")
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
else:
    print("Không có mô hình nào (ngoài Dummy) để vẽ.")

# 7.3. Residual Plot – mô hình tốt nhất (theo MAE)
best_model_name = results_df.iloc[0]["model"]
best_model = models[best_model_name]
best_pipe = Pipeline([("prep", preprocess), ("model", best_model)])
best_pipe.fit(X_train, y_train)
y_pred_best = best_pipe.predict(X_test)
residuals = y_test - y_pred_best
plt.figure(figsize=(10, 6))
indices = np.random.choice(len(y_pred_best), min(500, len(y_pred_best)), replace=False)
plt.scatter(y_pred_best[indices], residuals.iloc[indices], alpha=0.6, color="purple")
plt.axhline(0, color="red", linestyle="--", label="Residual = 0")
plt.xlabel("Predicted"); plt.ylabel("Residual (Actual - Predicted)")
plt.title(f"Residual Plot – Best Model: {best_model_name} (Đã sửa Leakage Triệt Để)")
plt.grid(True); plt.legend(); plt.tight_layout(); plt.show()
print(f"\n➡️ Mô hình tốt nhất (MAE thấp nhất): {best_model_name}")

# ============================================================
# 8. EXPLAINABILITY
# ============================================================
# ============================================================
# 8.1 SAVE FINAL MODEL – RANDOMFOREST_DEEP DÙNG CHO CELL 3
# ============================================================

import pickle

rf_model_name = "RandomForest_Deep"
if rf_model_name not in models:
    raise ValueError("Không tìm thấy RandomForest_Deep trong dict models.")

rf_model = models[rf_model_name]

# Pipeline cuối cùng: preprocess + RandomForest_Deep, train trên TOÀN BỘ DỮ LIỆU (X, y)
final_rf_pipe = Pipeline([
    ("prep", preprocess),
    ("model", rf_model)
])

print("\n===== TRAIN FINAL RANDOMFOREST_DEEP TRÊN TOÀN BỘ DỮ LIỆU =====")
final_rf_pipe.fit(X, y)

with open("model.pkl", "wb") as f:
    pickle.dump(final_rf_pipe, f)

print("🎉 ĐÃ LƯU PIPELINE RandomForest_Deep VÀO file: model.pkl")


# 8.2. Random Forest – feature_importances_
try:
    rf_model_name = "RandomForest_Deep"
    if rf_model_name not in models: rf_model_name = "RandomForest"

    # Tạo pipe CHỈ cho Random Forest
    rf_pipe = Pipeline([("prep", preprocess), ("model", models[rf_model_name])])
    rf_pipe.fit(X_train, y_train)

    # Lấy tên feature sau khi One-Hot-Encode TỪ BÊN TRONG PIPELINE
    if len(cat_features) > 0:
        ohe_rf = rf_pipe.named_steps["prep"].named_transformers_["cat"].named_steps["onehot"]
        cat_feature_names_rf = ohe_rf.get_feature_names_out(cat_features)
    else: 
        cat_feature_names_rf = []
        
    all_rf_features = num_features + list(cat_feature_names_rf)
    rf_importance = rf_pipe.named_steps["model"].feature_importances_
    
    imp_df = pd.DataFrame({"feature": all_rf_features, "importance": rf_importance})
    imp_df = imp_df.sort_values("importance", ascending=False)
    
    n_show_imp = min(20, imp_df.shape[0])
    print(f"\n===== TOP {n_show_imp} FEATURE IMPORTANCES – {rf_model_name} (Đã sửa Leakage Triệt Để) =====")
    display(imp_df.head(n_show_imp).style.background_gradient(cmap='Greens', subset=['importance']))

except Exception as e:
    print(f"Lỗi khi tính Feature Importance: {e}")

# ============================================================
# 9. (ĐÃ CHUYỂN LÊN MỤC 4B)
# ============================================================
# (Để trống mục này)
```

```python
# ============================================================
# CELL 3 – MULTIMODAL AI APP (OSM + RANDOMFOREST_DEEP)
#
# - Geocode: OpenStreetMap (Nominatim) + alias UEH/KTX
# - Dùng model RandomForest_Deep (model.pkl) đã train ở cell 2
# - Input model: 9 features giống hệt cell 2
# - traffic_level: random (low / medium / high) cho mỗi chuyến
#   → ảnh hưởng GIÁ qua:
#       + mô hình ML (đã train với traffic_level)
#       + TRAFFIC_MULTIPLIER cho phần rule-based fallback
# - User được chọn GrabBike / GrabCar (hệ số giá)
# - So sánh 3 phương án:
#       + Grab only
#       + Grab → Bus
#       + Grab → Metro (chỉ khi hợp lý)
# ============================================================

# Nếu môi trường chưa cài geopy thì chạy thêm:
# %pip install geopy

import pickle
import random
import time

import numpy as np
import pandas as pd

from geopy.geocoders import Nominatim
from geopy.distance import geodesic

# ============================================================
# 0. CONFIG: FILE MODEL & GIÁ CƯỚC DEMO
# ============================================================

MODEL_PATH = "model.pkl"  # pipeline RandomForest_Deep đã lưu từ V17

# Giá Grab demo (gần case 5km ~ 15p ~ 28k)
GRAB_BASE_FARE = 16000      # 2km đầu
GRAB_PER_KM   = 2600        # từ km thứ 3 trở đi
GRAB_PER_MIN  = 200         # mỗi phút di chuyển

BUS_FLAT_FARE   = 7000      # 1 lượt bus
METRO_FLAT_FARE = 15000     # 1 lượt metro

SPEED_GRAB  = 25            # km/h (demo)
SPEED_BUS   = 18            # km/h (demo)
SPEED_METRO = 30            # km/h (demo)

ETA_GRAB_DEFAULT = 5        # phút (Grab tới đón)
WAIT_BUS   = 8              # phút chờ bus
WAIT_METRO = 5              # phút chờ metro
WALK_MIN   = 10             # phút đi bộ (demo)

# Hệ số giá theo tình trạng giao thông
TRAFFIC_MULTIPLIER = {
    "low": 1.00,    # đường thoáng – không phụ phí
    "medium": 1.15, # hơi kẹt – tăng ~15%
    "high": 1.40    # kẹt nặng – tăng ~40%
}

# Ngưỡng khoảng cách tối thiểu để xem xét metro
MIN_METRO_DISTANCE_KM = 6.0   # có thể chỉnh xuống 4.0 hoặc 3.0 nếu muốn

# ============================================================
# 1B. DANH SÁCH TRẠM BUS / METRO DEMO
# ============================================================

bus_stops = [
    {
        "name": "Bến xe buýt Sài Gòn",
        "lat": 10.772510,
        "lon": 106.700993
    },
    {
        "name": "Trường Trần Khai Nguyên",
        "lat": 10.755350,
        "lon": 106.667990
    },
    {
        "name": "Bến xe buýt Chợ Lớn",
        "lat": 10.755564,
        "lon": 106.642331
    }
]

metro_stations = [
    {
        "name": "Ga Bến Thành",
        "lat": 10.772050,
        "lon": 106.698333
    },
    {
        "name": "Ga Khu Công Nghệ Cao",
        "lat": 10.850965,
        "lon": 106.802597
    },
    {
        "name": "Ga Suối Tiên",
        "lat": 10.872802,
        "lon": 106.807331
    }
]

# ============================================================
# 1. ALIAS ĐỊA CHỈ QUEN THUỘC + GEOCODE OSM (KHÔNG DÙNG GOOGLE)
# ============================================================

KNOWN_POINTS = {
    # ============================
    # UEH CƠ SỞ V – QUẬN 3
    # ============================
    "ueh cơ sở v": (
        10.786400,
        106.689200,
        "UEH – Cơ sở V, 232/6 Võ Thị Sáu, Quận 3, TP. Hồ Chí Minh, Việt Nam"
    ),
    "ueh co so v": (
        10.786400,
        106.689200,
        "UEH – Campus V, 232/6 Vo Thi Sau, District 3, Ho Chi Minh City, Vietnam"
    ),

    # ============================
    # UEH CƠ SỞ N – BÌNH CHÁNH
    # ============================
    "ueh cơ sở n": (
        10.740735,
        106.594235,
        "UEH – Cơ sở N, Khu phố 4, Phường Tân Nhựt, Huyện Bình Chánh, TP. Hồ Chí Minh, Việt Nam"
    ),
    "ueh co so n": (
        10.740735,
        106.594235,
        "UEH – Campus N, Tan Nhut Ward, Binh Chanh District, Ho Chi Minh City, Vietnam"
    ),

    # ============================
    # KTX UEH – NGUYỄN CHÍ THANH
    # ============================
    "ktx 43-45 nguyễn chí thanh": (
        10.757500,
        106.664000,
        "KTX UEH, 43–45 Nguyễn Chí Thanh, Quận 5, TP. Hồ Chí Minh, Việt Nam"
    ),
    "ktx 43 45 nguyễn chí thanh": (
        10.757500,
        106.664000,
        "KTX UEH, 43–45 Nguyễn Chí Thanh, Quận 5, TP. Hồ Chí Minh, Việt Nam"
    ),
    "ktx 43-45 nct": (
        10.757500,
        106.664000,
        "KTX UEH, 43–45 Nguyễn Chí Thanh, Quận 5, TP. Hồ Chí Minh, Việt Nam"
    ),
    "ktx 43 45 nct": (
        10.757500,
        106.664000,
        "KTX UEH, 43–45 Nguyễn Chí Thanh, Quận 5, TP. Hồ Chí Minh, Việt Nam"
    ),
    "ktx ueh": (
        10.757500,
        106.664000,
        "KTX UEH, 43–45 Nguyễn Chí Thanh, Quận 5, TP. Hồ Chí Minh, Việt Nam"
    ),
}


def normalize_address_input(text: str) -> str:
    """
    Chuẩn hóa địa chỉ nhập từ người dùng:
    - Nếu khớp alias quen thuộc → trả về key để lấy từ KNOWN_POINTS
    - Nếu không, tự động thêm 'Hồ Chí Minh, Việt Nam' nếu user không ghi thành phố
    """
    raw = text.strip().lower()
    if raw in KNOWN_POINTS:
        return raw
    if "hồ chí minh" not in raw and "ho chi minh" not in raw:
        raw = raw + ", Hồ Chí Minh, Việt Nam"
    return raw


def geocode_address(address: str):
    """
    Geocode địa chỉ → (lat, lon, full_address) bằng:
    - Alias UEH/KTX (KNOWN_POINTS), nếu trùng
    - Nominatim (OpenStreetMap) nếu không trùng
    """
    if not address or address.strip() == "":
        raise ValueError("Địa chỉ trống. Vui lòng nhập địa chỉ hợp lệ.")

    address_norm = normalize_address_input(address)
    key = address_norm.lower()

    # 1) Điểm quen thuộc
    if key in KNOWN_POINTS:
        lat, lon, full_addr = KNOWN_POINTS[key]
        return lat, lon, full_addr

    # 2) Địa chỉ thường → Nominatim
    geolocator = Nominatim(user_agent="hcm_multimodal_osm_demo")
    time.sleep(1)  # tránh spam server OSM
    loc = geolocator.geocode(address_norm)
    if not loc:
        raise ValueError(
            f"Nominatim không tìm được địa chỉ: {address_norm}. "
            f"Vui lòng thử lại với địa chỉ chi tiết hơn hoặc dùng alias UEH/KTX."
        )
    return loc.latitude, loc.longitude, loc.address

# ============================================================
# 2. HÀM TÍNH THỜI GIAN & GIÁ RULE-BASED
# ============================================================

def compute_grab_price(distance_km: float, duration_min: float) -> int:
    """Tính giá Grab-only theo bảng giá demo."""
    if distance_km <= 0:
        distance_km = 1.0
    extra_km = max(distance_km - 2.0, 0.0)
    total = (
        GRAB_BASE_FARE +
        extra_km * GRAB_PER_KM +
        max(duration_min, 0.0) * GRAB_PER_MIN
    )
    return round(total)


def estimate_grab_time(distance_km: float, eta_min: float = ETA_GRAB_DEFAULT) -> float:
    travel = distance_km / SPEED_GRAB * 60.0
    return travel + eta_min


def estimate_bus_time(distance_km: float) -> float:
    bus_travel = distance_km / SPEED_BUS * 60.0
    return bus_travel + WAIT_BUS + WALK_MIN


def estimate_metro_time(distance_km: float) -> float:
    metro_travel = distance_km / SPEED_METRO * 60.0
    return metro_travel + WAIT_METRO + WALK_MIN

# ============================================================
# 3. CHỌN LOẠI XE: GRABBIKE / GRABCAR (ÁP HỆ SỐ GIÁ)
# ============================================================

VEHICLE_LABEL = "GrabBike"
VEHICLE_PRICE_FACTOR = 1.0  # GrabBike = 1.0, GrabCar = 1.2 (ví dụ)

EHICLE_LABEL = "GrabCar"
VEHICLE_PRICE_FACTOR = 4.0
def ask_vehicle_type():
    """
    Hỏi người dùng muốn đi GrabBike hay GrabCar.
    Dùng hệ số giá đơn giản:
        - GrabBike: 1.0
        - GrabCar : 4
    """
    global VEHICLE_LABEL, VEHICLE_PRICE_FACTOR
    raw = input("Bạn muốn đi GrabBike hay GrabCar? (b/c, Enter = b): ").strip().lower()

    if raw == "c":
        VEHICLE_LABEL = "GrabCar"
        VEHICLE_PRICE_FACTOR = 4.0
    else:
        VEHICLE_LABEL = "GrabBike"
        VEHICLE_PRICE_FACTOR = 1.0

    print(f"→ Đã chọn loại xe: {VEHICLE_LABEL} (hệ số giá = {VEHICLE_PRICE_FACTOR})")

# ============================================================
# 4. RANDOM TRAFFIC_LEVEL + KHOẢNG CÁCH
# ============================================================

def random_traffic_level():
    """Mô phỏng tình trạng giao thông: low / medium / high."""
    return random.choice(["low", "medium", "high"])


def compute_distance_km(o_lat, o_lon, d_lat, d_lon, min_km=0.3):
    """
    Tính khoảng cách chim bay. Nếu < min_km thì ép lên min_km
    để tránh case 0 km khi hai điểm quá gần / trùng nhau.
    """
    raw = geodesic((o_lat, o_lon), (d_lat, d_lon)).km
    if raw < min_km:
        return min_km
    return raw

# ============================================================
# 4B. HÀM TÌM TRẠM GẦN NHẤT
# ============================================================

def find_nearest_stop(lat, lon, stops):
    """Tìm trạm gần nhất trong danh sách stops (bus hoặc metro)."""
    best_stop = None
    best_dist = float("inf")
    for s in stops:
        d = geodesic((lat, lon), (s["lat"], s["lon"])).km
        if d < best_dist:
            best_dist = d
            best_stop = s
    return best_stop, best_dist


def find_two_distinct_stops(o_lat, o_lon, d_lat, d_lon, stops):
    """
    Tìm 2 trạm khác nhau cho điểm đi và điểm đến.
    - pickup: trạm gần điểm đi nhất
    - dropoff: trạm gần điểm đến nhất, nhưng KHÁC tên với pickup
      nếu không tìm được trạm khác → trả về (pickup, None)
    """
    pickup, _ = find_nearest_stop(o_lat, o_lon, stops)
    dropoff, _ = find_nearest_stop(d_lat, d_lon, stops)

    if pickup is None or dropoff is None:
        return pickup, dropoff

    if pickup["name"] != dropoff["name"]:
        return pickup, dropoff

    # Nếu tên trạm trùng nhau → tìm trạm thứ 2 gần nhất cho điểm đến
    best2 = None
    best2_dist = float("inf")
    for s in stops:
        if s["name"] == pickup["name"]:
            continue
        d = geodesic((d_lat, d_lon), (s["lat"], s["lon"])).km
        if d < best2_dist:
            best2_dist = d
            best2 = s

    if best2 is None:
        # Không có trạm nào khác -> dropoff = None (bus không phù hợp)
        return pickup, None

    return pickup, best2

# ============================================================
# 5. LOAD PIPELINE RANDOMFOREST_DEEP
# ============================================================

try:
    with open(MODEL_PATH, "rb") as f:
        ml_model = pickle.load(f)
    print("✅ Đã load pipeline RandomForest_Deep từ model.pkl.")
except Exception as e:
    print(f"⚠ Không load được model.pkl ({e}) – sẽ chỉ dùng rule-based.")
    ml_model = None

# Đúng danh sách features mà V17 đã dùng để train
feature_cols_for_model = [
    "distance_km",
    "traffic_level",
    "bus_available",
    "metro_available",
    "passenger_count",
    "time_of_day_slot",
    "walking_distance_to_transit_m",
    "grab_eta_to_user_min",
    "nearest_driver_distance_km"
]

# ============================================================
# 6. BUILD 1 HÀNG FEATURES ĐÚNG FORMAT V17
# ============================================================

def build_feature_row(distance_km, mode_code, traffic_level):
    """
    Xây 1 DataFrame (1 dòng) với đầy đủ 9 cột features giống V17.
    mode_code:
        - "grab_only"
        - "grab_bus"
        - "grab_metro"
    traffic_level: "low" / "medium" / "high"
    """
    data = {}

    for col in feature_cols_for_model:
        if col == "distance_km":
            data[col] = distance_km

        elif col == "traffic_level":
            data[col] = traffic_level

        elif col == "bus_available":
            data[col] = 1 if mode_code == "grab_bus" else 0

        elif col == "metro_available":
            data[col] = 1 if mode_code == "grab_metro" else 0

        elif col == "passenger_count":
            data[col] = 1  # giả định 1 hành khách

        elif col == "time_of_day_slot":
            data[col] = "daytime"  # demo: ban ngày

        elif col == "walking_distance_to_transit_m":
            if mode_code == "grab_only":
                data[col] = 0
            elif mode_code == "grab_bus":
                data[col] = 500  # bus: đi bộ ~500m
            elif mode_code == "grab_metro":
                data[col] = 600  # metro: đi bộ ~600m
            else:
                data[col] = 0

        elif col == "grab_eta_to_user_min":
            data[col] = ETA_GRAB_DEFAULT

        elif col == "nearest_driver_distance_km":
            data[col] = 1.0  # giả định tài xế gần nhất cách 1km

        else:
            data[col] = 0

    return pd.DataFrame([data])

# ============================================================
# 7. DỰ ĐOÁN GIÁ BẰNG MODEL (CÓ FALLBACK RULE-BASED)
# ============================================================

def predict_price(distance_km, mode_code, rule_based_price, traffic_level):
    """
    Nếu có ml_model:
        - build DataFrame features
        - predict
      Nếu lỗi → fallback rule_based_price
    Nếu không có ml_model → dùng rule_based_price
    """
    if ml_model is None:
        return float(rule_based_price)

    try:
        X_row = build_feature_row(distance_km, mode_code, traffic_level)
        y_pred = ml_model.predict(X_row)
        return float(y_pred[0])
    except Exception as e:
        print(f"⚠ Lỗi khi predict với model cho mode={mode_code}: {e}")
        print("→ Dùng lại giá rule-based.")
        return float(rule_based_price)

# ============================================================
# 8. TÍNH CÁC PHƯƠNG ÁN MULTIMODAL
# ============================================================

def compute_multimodal_options(o_lat, o_lon, d_lat, d_lon):
    distance_km = compute_distance_km(o_lat, o_lon, d_lat, d_lon)

    # Chọn 1 traffic_level chung cho cả chuyến đi
    tl = random_traffic_level()
    print(f"\n→ traffic_level giả lập cho chuyến này: {tl}")

    options = []

    # 1) GRAB ONLY
    t_grab = estimate_grab_time(distance_km)
    rb_price_grab = compute_grab_price(distance_km, t_grab)
    rb_price_grab = rb_price_grab * TRAFFIC_MULTIPLIER[tl]
    price_grab = predict_price(distance_km, "grab_only", rb_price_grab, tl)

    options.append({
        "mode": "Grab only",
        "code": "grab_only",
        "distance_km": distance_km,
        "time_min": t_grab,
        "price_vnd": price_grab,
        "pickup_stop": None,
        "dropoff_stop": None
    })

    # 2) GRAB → BUS (chỉ nếu tìm được 2 trạm khác nhau)
    pickup_bus, dropoff_bus = find_two_distinct_stops(o_lat, o_lon, d_lat, d_lon, bus_stops)
    valid_bus = (pickup_bus is not None and dropoff_bus is not None)

    if valid_bus:
        t_bus = estimate_bus_time(distance_km)
        grab_part_bus = distance_km * 0.3
        rb_price_bus = compute_grab_price(grab_part_bus, t_bus * 0.3) + BUS_FLAT_FARE
        rb_price_bus = rb_price_bus * TRAFFIC_MULTIPLIER[tl]
        price_bus = predict_price(distance_km, "grab_bus", rb_price_bus, tl)

        options.append({
            "mode": "Grab → Bus",
            "code": "grab_bus",
            "distance_km": distance_km,
            "time_min": t_bus,
            "price_vnd": price_bus,
            "pickup_stop": pickup_bus,
            "dropoff_stop": dropoff_bus
        })

    # 3) GRAB → METRO (chỉ nếu khoảng cách đủ xa và ga lên/xuống khác nhau)
    pickup_metro, _ = find_nearest_stop(o_lat, o_lon, metro_stations)
    dropoff_metro, _ = find_nearest_stop(d_lat, d_lon, metro_stations)

    valid_metro = (
        pickup_metro is not None and
        dropoff_metro is not None and
        pickup_metro["name"] != dropoff_metro["name"] and
        distance_km >= MIN_METRO_DISTANCE_KM
    )

    if valid_metro:
        t_metro = estimate_metro_time(distance_km)
        grab_part_metro = distance_km * 0.2
        rb_price_metro = compute_grab_price(grab_part_metro, t_metro * 0.2) + METRO_FLAT_FARE
        rb_price_metro = rb_price_metro * TRAFFIC_MULTIPLIER[tl]
        price_metro = predict_price(distance_km, "grab_metro", rb_price_metro, tl)

        options.append({
            "mode": "Grab → Metro",
            "code": "grab_metro",
            "distance_km": distance_km,
            "time_min": t_metro,
            "price_vnd": price_metro,
            "pickup_stop": pickup_metro,
            "dropoff_stop": dropoff_metro
        })

    # Áp hệ số giá theo loại xe (GrabBike / GrabCar)
    for opt in options:
        opt["price_vnd"] = opt["price_vnd"] * VEHICLE_PRICE_FACTOR

    return options

# ============================================================
# 9. CHỌN PHƯƠNG ÁN TỐI ƯU
# ============================================================

def choose_best_option(options):
    # Ưu tiên rẻ nhất, nếu bằng thì nhanh hơn
    return min(options, key=lambda o: (o["price_vnd"], o["time_min"]))

# ============================================================
# 10. MAIN PIPELINE
# ============================================================

def run_pipeline_osm_rf():
    print("===== MULTIMODAL AI PIPELINE (OSM + RANDOMFOREST_DEEP) =====\n")

    # 1) Hỏi loại xe (GrabBike / GrabCar)
    ask_vehicle_type()

    # 2) Hỏi địa điểm
    dest = input("Bạn muốn đi đâu? ").strip()
    origin = input("Bạn đang ở đâu? ").strip()

    # 3) Geocode bằng OSM + alias
    print("\n→ Đang geocode (Nominatim + alias UEH/KTX)…")
    o_lat, o_lon, o_addr = geocode_address(origin)
    d_lat, d_lon, d_addr = geocode_address(dest)

    print(f"- Điểm đi : {o_addr} ({o_lat:.6f}, {o_lon:.6f})")
    print(f"- Điểm đến: {d_addr} ({d_lat:.6f}, {d_lon:.6f})")

    if abs(o_lat - d_lat) < 1e-5 and abs(o_lon - d_lon) < 1e-5:
        print("\n⚠ CẢNH BÁO: Điểm đi và điểm đến rất gần / trùng nhau.")
        print("→ Sẽ dùng khoảng cách tối thiểu 0.3 km để tính toán.\n")

    # 4) Tính các phương án
    print("→ Đang tính các phương án multimodal…")
    options = compute_multimodal_options(o_lat, o_lon, d_lat, d_lon)

    # 5) In kết quả từng phương án
    print("\n===== DANH SÁCH PHƯƠNG ÁN =====")
    for opt in options:
        print(f"\n🔸 {opt['mode']} ({VEHICLE_LABEL})")
        print(f"   • Quãng đường ước tính: {opt['distance_km']:.2f} km")
        print(f"   • Thời gian ước tính  : {opt['time_min']:.1f} phút")
        print(f"   • Chi phí ước tính    : {opt['price_vnd']:.0f} VND")

        # Mô tả trạm nếu có
        if opt["code"] == "grab_bus" and opt["pickup_stop"] and opt["dropoff_stop"]:
            print(f"   • Lộ trình bus       : Lên tại trạm '{opt['pickup_stop']['name']}', "
                  f"xuống tại trạm '{opt['dropoff_stop']['name']}'")
        if opt["code"] == "grab_metro" and opt["pickup_stop"] and opt["dropoff_stop"]:
            print(f"   • Lộ trình metro     : Lên tại ga '{opt['pickup_stop']['name']}', "
                  f"xuống tại ga '{opt['dropoff_stop']['name']}'")

    # 6) Chọn phương án tối ưu
    best = choose_best_option(options)

    print("\n===== PHƯƠNG ÁN TỐI ƯU =====")
    print(f"⭐ {best['mode']} ({VEHICLE_LABEL})")
    print(f"   • Thời gian: {best['time_min']:.1f} phút")
    print(f"   • Chi phí : {best['price_vnd']:.0f} VND")
    if best["code"] == "grab_bus" and best["pickup_stop"] and best["dropoff_stop"]:
        print(f"   • Chi tiết bus       : Lên tại trạm '{best['pickup_stop']['name']}', "
              f"xuống tại trạm '{best['dropoff_stop']['name']}'")
    if best["code"] == "grab_metro" and best["pickup_stop"] and best["dropoff_stop"]:
        print(f"   • Chi tiết metro     : Lên tại ga '{best['pickup_stop']['name']}', "
              f"xuống tại ga '{best['dropoff_stop']['name']}'")


# GỌI CHƯƠNG TRÌNH
run_pipeline_osm_rf()
```

```python
# ============================================================
# CELL 4 – XAI / MODEL EXPLAINABILITY WITH SHAP (RandomForest_Deep)
# - Giải thích pipeline RandomForest_Deep (ml_model) đã load từ model.pkl
# - Quy trình:
#     1. Sinh dữ liệu giả lập bằng build_feature_row(...)
#     2. Đưa qua bước "prep" trong pipeline để mã hoá/scale -> toàn số
#     3. Dùng TreeExplainer cho RandomForest (bước "model")
#     4. Vẽ:
#         - Global: summary bar + beeswarm
#         - Local : waterfall cho 1 kịch bản
# ============================================================

import shap
import matplotlib.pyplot as plt
import pandas as pd
import random

shap.initjs()

# ------------------------------------------------------------
# 1. TẠO DỮ LIỆU NỀN (BACKGROUND) ĐỂ GIẢI THÍCH
#    - Dùng đúng hàm build_feature_row(distance_km, mode_code, traffic_level)
#    - Các giá trị được random trong khoảng hợp lý
# ------------------------------------------------------------

def sample_background(n_samples=300):
    """
    Sinh n_samples dòng dữ liệu features đúng format V17
    để dùng cho SHAP (background).
    """
    rows = []
    for _ in range(n_samples):
        # Khoảng cách giả lập: 0.5km -> 20km
        distance_km = random.uniform(0.5, 20.0)

        # Mode: 3 phương án multimodal
        mode_code = random.choice(["grab_only", "grab_bus", "grab_metro"])

        # Giao thông: 3 mức
        traffic_level = random.choice(["low", "medium", "high"])

        row_df = build_feature_row(distance_km, mode_code, traffic_level)
        rows.append(row_df)

    bg = pd.concat(rows, ignore_index=True)
    return bg

X_background = sample_background(n_samples=300)
print("📌 Kích thước X_background (trước preprocess):", X_background.shape)
display(X_background.head())

# ------------------------------------------------------------
# 2. LẤY PREPROCESSOR & RANDOMFOREST TỪ PIPELINE ml_model
# ------------------------------------------------------------

if ml_model is None:
    print("⚠ ml_model đang là None – kiểm tra lại phần load model.pkl ở CELL 3.")
else:
    # Bước tiền xử lý trong pipeline (tên bước là "prep" như V17)
    try:
        prep = ml_model.named_steps["prep"]
    except KeyError:
        raise KeyError(
            "Không tìm thấy bước 'prep' trong ml_model.named_steps. "
            "Kiểm tra lại tên bước trong Pipeline (CELL 0/V17)."
        )

    # Bước mô hình RandomForest_Deep trong pipeline
    try:
        rf_model = ml_model.named_steps["model"]
    except KeyError:
        raise KeyError(
            "Không tìm thấy bước 'model' trong ml_model.named_steps. "
            "Kiểm tra lại tên bước mô hình trong Pipeline."
        )

    # --------------------------------------------------------
    # 3. TRANSFORM X_background -> TOÀN SỐ (CHO SHAP)
    #    - ColumnTransformer + OneHotEncoder sẽ sinh nhiều cột
    #    - Lấy tên cột mới bằng get_feature_names_out()
    # --------------------------------------------------------
    X_trans = prep.transform(X_background)
    feature_names = prep.get_feature_names_out()
    X_trans_df = pd.DataFrame(X_trans, columns=feature_names)

    print("📌 Kích thước X_trans_df (sau preprocess):", X_trans_df.shape)
    display(X_trans_df.head())

    # --------------------------------------------------------
    # 4. TẠO TREEEXPLAINER VÀ TÍNH GIÁ TRỊ SHAP
    # --------------------------------------------------------
    explainer = shap.TreeExplainer(rf_model)
    shap_values = explainer(X_trans_df)

    # --------------------------------------------------------
    # 5. GLOBAL EXPLANATION – FEATURE IMPORTANCE (BAR PLOT)
    # --------------------------------------------------------
    plt.figure()
    shap.summary_plot(
        shap_values,
        X_trans_df,
        plot_type="bar",
        show=False
    )
    plt.title("SHAP Feature Importance (Global) – RandomForest_Deep")
    plt.tight_layout()
    plt.show()

    # --------------------------------------------------------
    # 6. GLOBAL EXPLANATION – BEESWARM PLOT
    # --------------------------------------------------------
    plt.figure()
    shap.summary_plot(
        shap_values,
        X_trans_df,
        show=False
    )
    plt.title("SHAP Summary – Toàn bộ kịch bản (Beeswarm)")
    plt.tight_layout()
    plt.show()

    # --------------------------------------------------------
    # 7. LOCAL EXPLANATION – WATERFALL CHO 1 KỊCH BẢN CỤ THỂ
    # --------------------------------------------------------
    idx = 0  # đổi sang 1, 2, 10... nếu muốn xem các trường hợp khác
    print("🔍 Giải thích cho kịch bản có index (trong background):", idx)
    display(X_background.iloc[[idx]])  # hiển thị bản feature gốc để dễ hiểu

    shap.plots.waterfall(
        shap_values[idx],
        max_display=15
    )
```

```python
# ============================================================
# CELL 5 – BIAS & FAIRNESS AUDIT (FIXED: AUTO-RESTORE MODE_CODE)
# - Tự động load model & data.
# - Tự động khôi phục cột 'mode_code' bị thiếu để vẽ biểu đồ.
# ============================================================

import pickle
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os
import random

print("--- 🚀 BẮT ĐẦU QUY TRÌNH KIỂM TRA BIAS & FAIRNESS ---")

# 1. SAFETY CHECK: LOAD MODEL
# ------------------------------------------------------------
if 'ml_model' not in globals():
    print("⚠️ Chưa thấy 'ml_model'. Đang load từ file 'model.pkl'...")
    if os.path.exists('model.pkl'):
        with open('model.pkl', 'rb') as f:
            ml_model = pickle.load(f)
        print("✅ Đã load model thành công.")
    else:
        raise FileNotFoundError("❌ LỖI: Không tìm thấy file 'model.pkl'.")

# 2. SAFETY CHECK: TẠO DỮ LIỆU NẾU THIẾU
# ------------------------------------------------------------
if 'X_background' not in globals():
    print("⚠️ Chưa thấy 'X_background'. Đang tạo lại...")
    # (Đoạn code tạo lại dữ liệu phòng hờ)
    def sample_background_quick(n_samples=300):
        rows = []
        for _ in range(n_samples):
            # Random inputs
            distance_km = random.uniform(0.5, 20.0)
            mode_code = random.choice(["grab_only", "grab_bus", "grab_metro"])
            traffic_level = random.choice(["low", "medium", "high"])
            
            # Gọi hàm gốc của bạn
            row_df = build_feature_row(distance_km, mode_code, traffic_level)
            # Lưu lại mode_code gốc để tí nữa vẽ biểu đồ (QUAN TRỌNG)
            row_df["_original_mode_code"] = mode_code 
            rows.append(row_df)
        return pd.concat(rows, ignore_index=True)

    try:
        X_background = sample_background_quick(300)
        print("✅ Đã tạo xong dữ liệu.")
    except NameError:
        raise NameError("❌ Bạn cần chạy Cell chứa hàm 'build_feature_row' (thường là Cell 2) trước.")

# 3. DỰ ĐOÁN & KHÔI PHỤC CỘT 'mode_code'
# ------------------------------------------------------------
# Lọc bỏ cột tạm (nếu có) để model không bị lỗi thừa feature
cols_for_model = [c for c in X_background.columns if c != "_original_mode_code"]
X_for_pred = X_background[cols_for_model]

print(f"📊 Đang dự đoán trên {len(X_background)} dòng...")
y_pred_audit = ml_model.predict(X_for_pred)

# Tạo dataframe để audit
audit_df = X_background.copy()
audit_df["predicted_price"] = y_pred_audit

# --- ĐOẠN FIX LỖI QUAN TRỌNG NHẤT Ở ĐÂY ---
# Kiểm tra xem có cột mode_code chưa, nếu chưa thì tái tạo lại từ cột flag
if "mode_code" not in audit_df.columns:
    if "_original_mode_code" in audit_df.columns:
        # Nếu lúc tạo data ở trên đã lưu dự phòng
        audit_df["mode_code"] = audit_df["_original_mode_code"]
    else:
        # Nếu data đến từ Cell 4 (SHAP) và bị mất mode_code -> Tái tạo thủ công
        print("🛠 Đang khôi phục cột 'mode_code' từ dữ liệu có sẵn...")
        def restore_mode(row):
            # Logic: Nếu có metro -> là grab_metro, có bus -> grab_bus, ko có gì -> grab_only
            if row.get('metro_available', 0) == 1: return 'grab_metro'
            if row.get('bus_available', 0) == 1: return 'grab_bus'
            return 'grab_only'
        
        audit_df["mode_code"] = audit_df.apply(restore_mode, axis=1)

print("✅ Dữ liệu sẵn sàng để vẽ (đã có cột mode_code).")
display(audit_df[["mode_code", "predicted_price"]].head())

# 4. VẼ BIỂU ĐỒ (Đã an toàn)
# ------------------------------------------------------------

# Check 1: Mode Fairness
plt.figure(figsize=(10, 6))
sns.boxplot(data=audit_df, x="mode_code", y="predicted_price", palette="Set2")
plt.title("Phân phối giá dự đoán theo Phương tiện")
plt.ylabel("Giá (VND)")
plt.grid(True, alpha=0.3)
plt.show()

# Check 2: Traffic Sensitivity
plt.figure(figsize=(10, 6))
sns.violinplot(data=audit_df, x="traffic_level", y="predicted_price", 
               order=["low", "medium", "high"], palette="coolwarm")
plt.title("Độ nhạy giá theo Giao thông")
plt.ylabel("Giá (VND)")
plt.grid(True, alpha=0.3)
plt.show()

# 5. THỐNG KÊ
# ------------------------------------------------------------
print("\n🔢 GIÁ TRUNG BÌNH THEO NHÓM (VND):")
group_stats = audit_df.groupby(["mode_code", "traffic_level"])["predicted_price"].mean().unstack()
pd.options.display.float_format = '{:,.0f}'.format
display(group_stats)
```
